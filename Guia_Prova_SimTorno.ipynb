## 1.Resumo de Comandos (Apostila da Professora)

### Funções Especiais e de Posicionamento
* **# ( End of Block):** Finaliza o bloco ou sentença.
* **X / Z:** Eixo transversal (X) e eixo longitudinal (Z).
* **I / K:** Coordenadas de centro de arco (I para eixo X, K para eixo Z).
* **R:** Definição de raio.
* **N:** Identifica cada bloco ou sentença de informação.
* **T:** Seleciona as ferramentas na torre. A função `T00` cancela os corretores e determina o Ponto de Troca.
* **F:** Programa a velocidade de avanço para o trabalho em usinagem.
* **S:** Fornece o valor da velocidade de corte.
* **H:** Executa desvios incondicionais no programa.
* **P:** Identifica programas e sub-programas.
* **L:** Define o número de repetições de uma operação.
* **/** (Barra): Inibe a execução de blocos no programa.

###  Códigos Preparatórios (Funções G)
* **G00:** avanço rápido.
* **G01:** Interpolação linear com avanço programado.
* **G02:** Interpolação circular no sentido horário.
* **G03:** Interpolação circular no sentido anti-horário.
* **G04:** Tempo de permanência.
* **G20:** Programação em Diâmetro.
* **G21:** Programação em Polegada.
* **G40:** Cancela compensação do raio da ponta da ferramenta.
* **G41:** Compensação do raio da ferramenta à esquerda.
* **G42:** Compensação do raio da ferramenta à direita.
* **G53:** Cancela todos os Deslocamentos de Ponto Zero.
* **G54:** Ativa o primeiro Deslocamento de Ponto Zero.
* **G55:** Ativa o segundo Deslocamento de Ponto Zero.
* **G70:** Programação em Polegada.
* **G71:** Programação em Milímetro.
* **G73:** Interpolação Linear Ponto a ponto.
* **G90:** Programação em Coordenadas Absolutas.
* **G91:** Programação em Coordenadas Incrementais.
* **G92:** Definição de Origem temporária ou Limite de RPM.
* **G94:** Programação em Avanço por minuto.
* **G95:** Programação em Avanço por rotação.
* **G96:** Programação em Velocidade de Corte Constante.
* **G97:** Programação em RPM direta.
* **G99:** Cancela Definição de Origem Temporária.

###  Códigos Miscelâneos (Funções M)
* **M00:** Parada programada.
* **M02:** Fim de programa.
* **M03:** Sentido horário de rotação do eixo-árvore.
* **M04:** Sentido anti-horário de rotação do eixo-árvore.
* **M05:** Desliga o eixo-árvore sem orientação.
* **M06:** Libera o giro da Torre na troca automática de ferramentas.
* **M07:** Liga refrigerante de corte de alta pressão.
* **M08:** Liga refrigerante de corte de baixa pressão.
* **M09:** Desliga refrigerante de corte.
* **M10 a M14:** Faixas de Rotação.
* **M24:** Abre a Placa de Fixação.
* **M25:** Fecha a Placa de Fixação.
* **M26:** Recua o Mangote da contra ponta.
* **M27:** Avança o Mangote da contra ponta.
* **M30:** Fim de programa.

### Função G66 (Ciclo Automático de Desbaste Longitudinal)
Este ciclo permite a usinagem de desbaste completa da peça, utilizando-se apenas de um bloco de programação, contendo apenas os parâmetros necessários para sua execução.

**Sintaxe da função:** `NXX G66 X... Z... I... K... ( U1 ) W... P... F... #`

**Onde:**
* **X:** Diâmetro de referência para início de torneamento (vide regra de posicionamento)
* **Z:** Comprimento de referência para início de torneamento (vide regra de posicionamento)
* **I:** Sobremetal para acabamento no eixo X
* **K:** Sobremetal para acabamento no eixo Z
* **(U1):** Pré-alisamento paralelo ao perfil final, mantendo as dimensões pré-estabelecidas
* **W:** Incremento por passada no diâmetro
* **P:** Subprograma que contém as dimensões de acabamento do perfil final da peça
* **F:** Avanço programado para desbaste

**Exemplo:**
`N50 G66 X75. Z82 I1. K.2 U1 W3. P10 F.2 #`

**Notas sobre G66:**
* Na função G66 o subprograma não aceita inversões de cotas nos eixos "X" e "Z".
* A função G66 não é Modal e requer um subprograma com as dimensões de acabamento da peça.
* Sempre o último valor de ‘X’ do subprograma (Externo ou Interno), deverá informar o diâmetro bruto do material, no caso de furos informar seu diâmetro.

---

* **K:** Sobremetal para acabamento no eixo Z

### Regra de Posicionamento para os eixos X e Z (G66 e G67)
A regra para posicionamento inicial do ciclo de desbaste deverá seguir as seguintes condições:

**Externo:**
* **X =** Maior diâmetro da peça em bruto + 4 mm
* **Z =** Comprimento da peça em bruto + 2 mm

**Interno:**
* **X =** Menor diâmetro da peça em bruto - 4 mm
* **Z =** Comprimento da peça em bruto + 2 mm

* Para utilizarmos o mesmo subprograma de desbaste no acabamento da peça, utilizando-se ferramentas diferentes, será necessário que ambas estejam no mesmo quadrante.
* As funções "G" admissíveis no subprograma são G01, G02, G03, G04 e G73.
